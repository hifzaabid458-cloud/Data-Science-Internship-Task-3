## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
import pandas as pd

df = pd.read_excel("Online Retail.xlsx")

df.head()
```

```python
print("Dataset Shape:", df.shape)
df.info()
```

## 3. Data Inspection

```python
df.describe()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

## 4. Data Cleaning

```python
df = df.dropna(subset=['CustomerID'])

print("Shape after removing missing CustomerID:", df.shape)
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
df = df[df['Quantity'] > 0]
df = df[df['UnitPrice'] > 0]

print("Shape after cleaning:", df.shape)
```

## 5. Create Customer-Level Features

```python
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']

df[['Quantity', 'UnitPrice', 'TotalAmount']].head()
```

### RFM Analysis

RFM analysis is used to summarize customer purchasing behaviour using three measures:

- **Recency** – How recently a customer made a purchase.
- **Frequency** – How often a customer made purchases.
- **Monetary** – How much a customer spent.

```python
reference_date = df['InvoiceDate'].max() + pd.Timedelta(days=1)

print("Reference Date:", reference_date)
```

```python
rfm = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (reference_date - x.max()).days,
    'InvoiceNo': 'nunique',
    'TotalAmount': 'sum'
})

rfm.columns = ['Recency', 'Frequency', 'Monetary']

rfm.head()
```

```python
print("Customer Segmentation Dataset Shape:", rfm.shape)
rfm.describe()
```

## 6. Feature Scaling

K-Means clustering is distance-based, so the RFM features need to be standardized before clustering. Standardization puts Recency, Frequency, and Monetary on a comparable scale.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm)

rfm_scaled = pd.DataFrame(
    rfm_scaled,
    columns=rfm.columns,
    index=rfm.index
)

rfm_scaled.head()
```

## 7. Finding the Optimal Number of Clusters

The Elbow Method is used to determine a suitable number of customer clusters. The Within-Cluster Sum of Squares (WCSS) is calculated for different values of K.

```python
wcss = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(rfm_scaled)
    wcss.append(kmeans.inertia_)

print("WCSS values:", wcss)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(range(2, 11), wcss, marker='o')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.xticks(range(2, 11))
plt.show()
```

## 8. K-Means Clustering

Based on the Elbow Method, 4 clusters are selected for the initial customer segmentation. K-Means will group customers according to similarities in their Recency, Frequency, and Monetary behaviour.

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

rfm['Cluster'] = kmeans.fit_predict(rfm_scaled)

rfm.head()
```

## 9. Silhouette Score

The Silhouette Score is used to evaluate how well customers are separated into their assigned clusters. A higher score generally indicates better-defined clusters.

```python
from sklearn.metrics import silhouette_score

sil_score = silhouette_score(rfm_scaled, rfm['Cluster'])

print("Silhouette Score:", sil_score)
```

```python
silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(rfm_scaled)
    score = silhouette_score(rfm_scaled, labels)
    silhouette_scores.append(score)

print("Silhouette Scores:", silhouette_scores)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.title('Silhouette Score for Different Numbers of Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.xticks(range(2, 11))
plt.show()
```

## 10. Final K-Means Model

The Silhouette Score comparison shows that **K = 2** provides the highest score of **0.8958**, indicating the strongest separation between customer groups.

Therefore, 2 clusters are selected as the final K-Means solution.

```python
final_kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

rfm['Cluster'] = final_kmeans.fit_predict(rfm_scaled)

print("Customers in Each Final Cluster:")
print(rfm['Cluster'].value_counts().sort_index())
```

```python
final_silhouette = silhouette_score(
    rfm_scaled,
    rfm['Cluster']
)

print("Final Silhouette Score:", final_silhouette)
```

## 11. Customer Segment Analysis

The characteristics of each customer segment will be analyzed using the average Recency, Frequency, and Monetary values. This will help identify differences in customer purchasing behaviour between the clusters.

```python
cluster_profile = rfm.groupby('Cluster')[[
    'Recency',
    'Frequency',
    'Monetary'
]].mean()

cluster_profile
```

## 12. Customer Segment Visualization

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    rfm['Frequency'],
    rfm['Monetary'],
    c=rfm['Cluster'],
    alpha=0.6
)

plt.title('Customer Segmentation')
plt.xlabel('Frequency')
plt.ylabel('Monetary Value')
plt.show()
```

### Cluster Characteristics

The customer segments can be characterized based on their average Recency, Frequency, and Monetary values.

- **Cluster 0 – Regular/Lower-Value Customers:** These customers have relatively higher recency, lower purchase frequency, and considerably lower monetary value.
- **Cluster 1 – High-Value Customers:** These customers purchased very recently, made purchases much more frequently, and generated substantially higher monetary value.

```python
cluster_summary = rfm.groupby('Cluster').agg(
    Customers=('Cluster', 'count'),
    Avg_Recency=('Recency', 'mean'),
    Avg_Frequency=('Frequency', 'mean'),
    Avg_Monetary=('Monetary', 'mean')
).round(2)

cluster_summary
```

## 13. Conclusion

K-Means clustering was successfully applied to segment customers based on their purchasing behaviour. RFM analysis was used to create customer-level features, and the final two-cluster solution achieved a Silhouette Score of 0.8958.

The resulting segments show a clear difference between regular/lower-value customers and a small group of highly active, high-value customers.

## 14. Cluster Comparison

```python
cluster_summary[['Avg_Recency', 'Avg_Frequency', 'Avg_Monetary']].plot(
    kind='bar',
    figsize=(10, 6)
)

plt.title('Comparison of Customer Segments')
plt.xlabel('Cluster')
plt.ylabel('Average Value')
plt.xticks(rotation=0)
plt.legend(title='RFM Metrics')
plt.show()
```
